```python
import asyncio
from google import genai
import os
from dotenv import load_dotenv

load_dotenv()

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)


async def agent(name, task):

    prompt = f"""
    You are the {name}.

    Task:
    {task}
    """

    response = await asyncio.to_thread(
        client.models.generate_content,
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text


async def main():

    topic = "AI agents"

    results = await asyncio.gather(

        agent(
            "technical researcher",
            f"Study the technical architecture of {topic}"
        ),

        agent(
            "business researcher",
            f"Study business applications of {topic}"
        ),

        agent(
            "security researcher",
            f"Study security risks of {topic}"
        )
    )

    print("===== TECHNICAL =====")
    print(results[0])

    print("\n===== BUSINESS =====")
    print(results[1])

    print("\n===== SECURITY =====")
    print(results[2])


await main()
```
