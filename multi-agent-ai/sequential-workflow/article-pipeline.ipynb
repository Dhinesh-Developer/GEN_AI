```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

def ask_gemini(prompt):
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )
    return response.text
```

```python
def researcher(topic):
    return ask_gemini(
        f"Research {topic} and give 10 important facts"
    )

def writer(research):
    return ask_gemini(
        f"write an article using the research: \n{research}"
    )

def editor(article):
    return ask_gemini(
        f"Improve this article \n {article}"
    )

topic = "Generative AI"

research = researcher(topic)
article = writer(research)
final_artical = editor(article)

print(final_artical)
```
