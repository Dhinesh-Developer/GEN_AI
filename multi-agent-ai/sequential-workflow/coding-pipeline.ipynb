```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

def ask_gemini(prompt):
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )
    return response.text
```

```python
def plan(requirement):
    return ask_gemini(
        f"Create an implementation plan:\n{requirement}"
    )

def developer(plan):
    return ask_gemini(
        f"Write python code based on this plan:\n{plan}"
    )

def tester(code):
    return ask_gemini(
        f"Create tests for this code:\n {code}"
    )

def reviewer(code, tests):
    return ask_gemini(
        f"""
Review this implementation:

CODE: {code}
TESTS: {tests}
"""
    )

requirement = "Create a calculator API."
plan = plan(requirement)
code = developer(plan)
tests = tester(code)
review = reviewer(code, tests)

print(review)


```
