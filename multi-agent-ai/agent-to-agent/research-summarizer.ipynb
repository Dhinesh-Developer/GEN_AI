```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

def ask_gemini(prompt):
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )
    return response.text
```

```python
def researcher(topic):
    prompt = f"""
You are a research agent.
Research the following topic from your knowledge:
{topic}
Give 5 important points.
"""
    return ask_gemini(prompt=prompt)

def summarizer(research):
    prompt = f"""
You are a summarization agent.
Summarize the following research:
{research}
Give a simple explanation suitable for a beginner.
"""
    return ask_gemini(prompt=prompt)

topic = "What is Deep Agent"

research = researcher(topic)
summary = summarizer(research=research)

print("RESEARCH")
print(research)

print("SUMMARY")
print(summary)
```
