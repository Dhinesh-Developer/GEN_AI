```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

def ask_gemini(prompt):
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )
    return response.text
```

```python
def planner(goal):
    prompt = f"""
You are a planning agent.
Goal:
{goal}

Break the goal into 5 pratical steps
"""
    return ask_gemini(prompt)

def executor(plan):
    prompt = f"""
Yoou are an execution agent.
Execute the following plan accordingly:
{plan}

Explain what should be done at each step.
"""
    return ask_gemini(prompt)

goal = "5 days japan visit, i am from india"
plan = planner(goal)
res = executor(plan)

print("===PLAN===")
print(plan)

print("===Executor===")
print(res)

```
