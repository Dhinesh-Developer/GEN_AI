3. Subtopic 1 — Agent

Before multiple agents, you need to understand what an agent actually is.

A simple agent can be represented as:

Agent
 ├── Role
 ├── Goal
 ├── Instructions
 ├── LLM
 └── Output

```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

def ask_gemini(prompt):
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )
    return response.text
```

```python
def main():
    prompt = """
You are a python teacher.

Explain python lists to a beginner.
Use simple examples
"""
    ans = ask_gemini(prompt=prompt)
    print(ans)


if __name__ == "__main__":
    main()

```
