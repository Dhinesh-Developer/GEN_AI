```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

def ask_gemini(prompt):
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )
    return response.text
```

```python
def supervisor(request):
    prompt = f"""
You are a supervisor.
Decide which specialist should handle this request.
Available agents:

RESEARCH
CODING
TESTING

Request: {request}

Return only one:
RESEARCH
CODING
TESTING
"""
    return ask_gemini(prompt).strip().upper()
```

```python
def research_agent(request):
    return ask_gemini(
        f"Research this topic:\n{request}"
    )

def coding_agent(request):
    return ask_gemini(
        f"Write python code for:\n{request}"
    )

def testing_agent(request):
    return ask_gemini(
        f"Create test cases for:\n{request}"
    )
```

```python
def run(request):
    decision = supervisor(request)
    print("supervisor:",decision)

    if "RESEARCH" in decision:
        return research_agent(request)
    elif "CODING" in decision:
        return coding_agent(request)
    elif "TESTING" in decision:
        return testing_agent(request)
    else:
        return "No suitable agent found."    
```

```python
request = input("Enter request:")
res = run(request)

print("\nRESULT:")
print(res)
```
