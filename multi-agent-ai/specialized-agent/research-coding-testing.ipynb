5. Subtopic 3 — Specialized Agents

This is where multi-agent architecture becomes useful.

Instead of:

Agent
does everything

you create:

Research Agent
Coding Agent
Testing Agent
Review Agent

Each has a narrow responsibility.

```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

def ask_gemini(prompt):
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )
    return response.text
```

```python
def research_agent(topic):
    prompt = f"""
You are a research specialist.

Research: {topic}

Give the important technical requirements
"""
    return ask_gemini(prompt)

def coding_agent(requirements):
    prompt = f"""
You are a python developer.
Requirements: {requirements}
write python code
"""
    return ask_gemini(prompt)

def testing_agent(code):
    prompt = f"""
You are a QA engineer.
Review the following python code and create test cases.
Code: {code}
"""

    return ask_gemini(prompt)

topic = "Python rest api"

research = research_agent(topic)
code = coding_agent(research)
tests = testing_agent(code)

print("RESEARCH")
print(research)

print("\nCODE")
print(code)

print("\nTEST CASES")
print(tests)
```
